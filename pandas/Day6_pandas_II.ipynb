```python
lg = pd.read_csv("logins.csv", parse_dates=["login_date"])   # read as real dates, not text

lg = lg.sort_values(["emp_id", "login_date"])                 # ← the ORDER BY
lg["prev_login"] = lg.groupby("emp_id")["login_date"].shift(1) # ← LAG, PARTITION BY emp_id
lg["gap_days"]   = (lg["login_date"] - lg["prev_login"]).dt.days
```

```python
bands = pd.read_csv("salary_bands.csv")
bands.shape
```

```python
e = emp.merge(dept, on="dept_id", how="left")
e["dept_name"] = e["dept_name"].fillna("Unknown")
e.shape
```

```python
x = e.merge(bands, how="cross")
x
```

```python
x = x[x["salary"].between(x["min_salary"], x["max_salary"])]
x
```

```python
x["rank_in_band"] = (x.groupby("band_name")["salary"]
                      .rank(method="dense", ascending=False)
                      .astype(int))

result = (x.sort_values(["band_name", "rank_in_band"])
           [["emp_name", "dept_name", "salary", "band_name", "rank_in_band"]])
result
```
